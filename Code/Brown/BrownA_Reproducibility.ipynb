```python
## Section 1 

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import accuracy_score, classification_report

from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.naive_bayes import GaussianNB

### import needed libraries
```

```python
## Section 2

xuetangx_df = pd.read_csv("~/Desktop/DATA 6550/all_normalized_features.csv")

xuetangx_df.head()

### display dataset 
```

```python
## Section 3

xuetangx_X = xuetangx_df.drop(columns=['truth', 'enroll_id'])
xuetangx_y = xuetangx_df['truth']

### features and target (truth) were separated from the XuetangX dataset
```

```python
## Section 4

xuetangx_X_train, xuetangx_X_test, xuetangx_y_train, xuetangx_y_test = train_test_split(
    xuetangx_X, xuetangx_y, test_size=0.2, random_state=100)

### the XuetangX data was split into training and testing sets with an 80/20 ratio
```

```python
## Section 5

x_sample = xuetangx_X_train.sample(n=100, random_state=100)
y_sample = xuetangx_y_train.loc[x_sample.index]

### sample of 100 training records was selected for lightweight model training and testing
```

```python
## Section 6

rf_model = RandomForestClassifier(n_estimators=10, random_state=100)

rf_model.fit(x_sample, y_sample)
rf_pred = rf_model.predict(xuetangx_X_test)

print("RF Accuracy:", accuracy_score(xuetangx_y_test, rf_pred))
print("\n", classification_report(xuetangx_y_test, rf_pred))

rf_cv = cross_val_score(rf_model, x_sample, y_sample, cv=5, n_jobs=1)
print("RF CV Mean:", np.mean(rf_cv))
print("RF CV Std:", np.std(rf_cv))

### Random Forest model trained on the 100-sample subset and evaluated using test accuracy and 5-fold cross-validation
```

### The Random Forest model achieved 77.3% accuracy on the test set, with strong performance on class 1 (F1-score: 0.85). Cross-validation showed a mean accuracy of 78% with a standard deviation of ~4%, indicating consistent performance.

```python
## Section 7

ab_model = AdaBoostClassifier(n_estimators=10, random_state=100)

ab_model.fit(x_sample, y_sample)
ab_pred = ab_model.predict(xuetangx_X_test)

print("AdaBoost Accuracy:", accuracy_score(xuetangx_y_test, ab_pred))
print("\n", classification_report(xuetangx_y_test, ab_pred))

ab_cv = cross_val_score(ab_model, x_sample, y_sample, cv=5, n_jobs=1)
print("AdaBoost CV Mean:", np.mean(ab_cv))
print("AdaBoost CV Std:", np.std(ab_cv))

### An AdaBoost model was trained on the 100-sample subset and evaluated using test accuracy and 5-fold cross-validation
```

### The AdaBoost model achieved 76.3% test accuracy, performing well on class 1 (F1-score: 0.85) but weaker on class 0. Cross-validation showed a mean accuracy of 78% with a low standard deviation of ~2.4%, indicating stable performance.

```python
## Section 8

dt_model = DecisionTreeClassifier(max_depth=3, random_state=100)

dt_model.fit(x_sample, y_sample)
dt_pred = dt_model.predict(xuetangx_X_test)

print("Decision Tree Accuracy:", accuracy_score(xuetangx_y_test, dt_pred))
print("\n", classification_report(xuetangx_y_test, dt_pred))

dt_cv = cross_val_score(dt_model, x_sample, y_sample, cv=5, n_jobs=1)
print("Decision Tree CV Mean:", np.mean(dt_cv))
print("Decision Tree CV Std:", np.std(dt_cv))

### A Decision Tree model with max depth 3 was trained on the 100-sample subset and evaluated using test accuracy and 5-fold cross-validation.
```

### The Decision Tree model achieved 76.5% test accuracy, with strong performance on class 1 (F1-score: 0.85) but limited recall for class 0. Cross-validation gave a 77% mean accuracy with a higher standard deviation of ~9.8%, indicating more variability across folds.

```python
## Section 9

knn_model = KNeighborsClassifier(n_neighbors=3)

knn_model.fit(x_sample, y_sample)
knn_pred = knn_model.predict(xuetangx_X_test)

print("kNN Accuracy:", accuracy_score(xuetangx_y_test, knn_pred))
print("\n", classification_report(xuetangx_y_test, knn_pred))

knn_cv = cross_val_score(knn_model, x_sample, y_sample, cv=5, n_jobs=1)
print("kNN CV Mean:", np.mean(knn_cv))
print("kNN CV Std:", np.std(knn_cv))

### A k-Nearest Neighbors (kNN) model with 3 neighbors was trained on the 100-sample subset and evaluated using test accuracy and 5-fold cross-validation.
```

### The kNN model achieved 77.2% test accuracy, performing very well on class 1 (F1-score: 0.86) but poorly on class 0 (F1-score: 0.30). Cross-validation showed a strong mean accuracy of 83% with a low standard deviation of ~2.4%, indicating consistent performance.

```python
## Section 10

lr_model = LogisticRegression(max_iter=1000, random_state=100)

lr_model.fit(x_sample, y_sample)
lr_pred = lr_model.predict(xuetangx_X_test)

print("LogReg Accuracy:", accuracy_score(xuetangx_y_test, lr_pred))
print("\n", classification_report(xuetangx_y_test, lr_pred))

lr_cv = cross_val_score(lr_model, x_sample, y_sample, cv=5, n_jobs=1)
print("LogReg CV Mean:", np.mean(lr_cv))
print("LogReg CV Std:", np.std(lr_cv))

### A Logistic Regression model was trained on the 100-sample subset with 1000 max iterations and evaluated using test accuracy and 5-fold cross-validation.
```

### The Logistic Regression model achieved 79.3% test accuracy, with excellent recall for class 1 (0.97) but low recall for class 0 (0.24). Cross-validation yielded a mean accuracy of 77% and a standard deviation of ~4%, showing stable performance.

```python
## Section 11

lda_model = LinearDiscriminantAnalysis()

lda_model.fit(x_sample, y_sample)
lda_pred = lda_model.predict(xuetangx_X_test)

print("LDA Accuracy:", accuracy_score(xuetangx_y_test, lda_pred))
print("\n", classification_report(xuetangx_y_test, lda_pred))

lda_cv = cross_val_score(lda_model, x_sample, y_sample, cv=5, n_jobs=1)
print("LDA CV Mean:", np.mean(lda_cv))
print("LDA CV Std:", np.std(lda_cv))

### A Linear Discriminant Analysis (LDA) model was trained on the 100-sample subset and evaluated using test accuracy and 5-fold cross-validation.
```

### The LDA model achieved 77.5% test accuracy, performing strongly on class 1 (F1-score: 0.86) but showing limited performance on class 0. Cross-validation resulted in a mean accuracy of 81% with a standard deviation of ~5.8%, indicating moderate variability across folds.

```python
## Section 12

nb_model = GaussianNB()

nb_model.fit(x_sample, y_sample)
nb_pred = nb_model.predict(xuetangx_X_test)

print("Naive Bayes Accuracy:", accuracy_score(xuetangx_y_test, nb_pred))
print("\n", classification_report(xuetangx_y_test, nb_pred))

nb_cv = cross_val_score(nb_model, x_sample, y_sample, cv=5, n_jobs=1)
print("NB CV Mean:", np.mean(nb_cv))
print("NB CV Std:", np.std(nb_cv))

### A Gaussian Naive Bayes model was trained on the 100-sample subset and evaluated using test accuracy and 5-fold cross-validation.
```

### The Naive Bayes model achieved 78.4% test accuracy, with solid performance on class 1 (F1-score: 0.86) and improved performance on class 0 compared to other models (F1-score: 0.51). Cross-validation produced a mean accuracy of 77% with a standard deviation of ~8.1%, indicating slightly higher variability across folds.

```python
## Section 13

results = {
    'Model': [
        'Random Forest', 'AdaBoost', 'Decision Tree', 'kNN',
        'Logistic Regression', 'LDA', 'Naive Bayes'
    ],
    'Test Accuracy': [
        accuracy_score(xuetangx_y_test, rf_pred),
        accuracy_score(xuetangx_y_test, ab_pred),
        accuracy_score(xuetangx_y_test, dt_pred),
        accuracy_score(xuetangx_y_test, knn_pred),
        accuracy_score(xuetangx_y_test, lr_pred),
        accuracy_score(xuetangx_y_test, lda_pred),
        accuracy_score(xuetangx_y_test, nb_pred)
    ],
    'CV Mean Accuracy': [
        np.mean(rf_cv), np.mean(ab_cv), np.mean(dt_cv),
        np.mean(knn_cv), np.mean(lr_cv), np.mean(lda_cv),
        np.mean(nb_cv)
    ],
    'CV Std Dev': [
        np.std(rf_cv), np.std(ab_cv), np.std(dt_cv),
        np.std(knn_cv), np.std(lr_cv), np.std(lda_cv),
        np.std(nb_cv)
    ]
}

results_df = pd.DataFrame(results)
results_df

### This code creates a DataFrame that compares the performance of seven ML models on the XuetangX dataset by compiling their test accuracy, cross-validation mean accuracy, and cross-validation standard deviation into a results table.
```

### The results show that all seven models achieved similar test accuracies around 76–79%, with Logistic Regression having the highest test accuracy (79.3%) and kNN achieving the highest cross-validation mean accuracy (83%), while Decision Tree and Naive Bayes showed more variability across folds.

### Overall Summary 

This project focused on reproducing machine learning results from a research paper that predicted student dropout using the XuetangX dataset. I used a preprocessed file called all_normalized_features.csv, where I separated the features from the target column, truth.

The data was split into training and testing sets (80/20). Because of system limitations, I sampled only 100 rows from the training set to avoid memory issues and crashing while still following the paper’s process.

I trained seven machine learning models: Random Forest, AdaBoost, Decision Tree, kNN, Logistic Regression, LDA, and Naive Bayes. Each model was evaluated using test accuracy, classification metrics, and 5-fold cross-validation.

Logistic Regression had the highest test accuracy (79.3%), while kNN had the best CV accuracy (83%). Most models performed well on class 1 (dropouts) but not as well on class 0 (non-dropouts), likely due to class imbalance.

Even though I trained on a smaller sample, the results closely followed the trends in the original paper. The small differences highlight how tough it is to fully reproduce results — especially when working with limited resources.

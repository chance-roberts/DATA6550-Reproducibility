```python
import pandas as pd

df = pd.read_csv("~/Desktop/DATA 6550/all_normalized_features.csv")

df.head()
```

```python
X = df.drop(columns=['truth', 'enroll_id'])
y = df['truth']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=100)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

rf_model = RandomForestClassifier(random_state=100)

rf_model.fit(X_train, y_train)

y_pred = rf_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
from sklearn.model_selection import cross_val_score
import numpy as np

cv_scores = cross_val_score(rf_model, X, y, cv=5)

print("CV Accuracy Scores:", cv_scores)
print("Mean Accuracy:", np.mean(cv_scores))
print("Standard Deviation:", np.std(cv_scores))
```

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(use_label_encoder=False, eval_metric='logloss', random_state=100)

xgb_model.fit(X_train, y_train)

xgb_pred = xgb_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, xgb_pred))
print("\nClassification Report:\n", classification_report(y_test, xgb_pred))
```

```python
xgb_cv_scores = cross_val_score(xgb_model, X, y, cv=5)

print("CV Accuracy Scores:", xgb_cv_scores)
print("Mean Accuracy:", np.mean(xgb_cv_scores))
print("Standard Deviation:", np.std(xgb_cv_scores))
```

```python
from sklearn.ensemble import AdaBoostClassifier

ab_model = AdaBoostClassifier(random_state=100)

ab_model.fit(X_train, y_train)

ab_pred = ab_model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, ab_pred))
print("\nClassification Report:\n", classification_report(y_test, ab_pred))
```

```python
ab_cv_scores = cross_val_score(ab_model, X, y, cv=5)

print("CV Accuracy Scores:", ab_cv_scores)
print("Mean Accuracy:", np.mean(ab_cv_scores))
print("Standard Deviation:", np.std(ab_cv_scores))
```

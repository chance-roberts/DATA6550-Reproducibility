```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.metrics import accuracy_score, classification_report

# Models
from sklearn.ensemble import RandomForestClassifier, AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.naive_bayes import GaussianNB
```

```python
# Load your dataset
xuetangx_df = pd.read_csv("~/Desktop/DATA 6550/all_normalized_features.csv")

# Preview
xuetangx_df.head()
```

```python
# Isolate X and y features
xuetangx_X = xuetangx_df.drop(columns=['truth', 'enroll_id'])
xuetangx_y = xuetangx_df['truth']
```

```python
xuetangx_X_train, xuetangx_X_test, xuetangx_y_train, xuetangx_y_test = train_test_split(
    xuetangx_X, xuetangx_y, test_size=0.2, random_state=100)
```

```python
# For resource-safe training & CV
x_sample = xuetangx_X_train.sample(n=100, random_state=100)
y_sample = xuetangx_y_train.loc[x_sample.index]
```

```python
rf_model = RandomForestClassifier(n_estimators=10, random_state=100)

rf_model.fit(x_sample, y_sample)
rf_pred = rf_model.predict(xuetangx_X_test)

print("RF Accuracy:", accuracy_score(xuetangx_y_test, rf_pred))
print("\n", classification_report(xuetangx_y_test, rf_pred))

rf_cv = cross_val_score(rf_model, x_sample, y_sample, cv=5, n_jobs=1)
print("RF CV Mean:", np.mean(rf_cv))
print("RF CV Std:", np.std(rf_cv))
```

```python
ab_model = AdaBoostClassifier(n_estimators=10, random_state=100)

ab_model.fit(x_sample, y_sample)
ab_pred = ab_model.predict(xuetangx_X_test)

print("AdaBoost Accuracy:", accuracy_score(xuetangx_y_test, ab_pred))
print("\n", classification_report(xuetangx_y_test, ab_pred))

ab_cv = cross_val_score(ab_model, x_sample, y_sample, cv=5, n_jobs=1)
print("AdaBoost CV Mean:", np.mean(ab_cv))
print("AdaBoost CV Std:", np.std(ab_cv))
```

```python
dt_model = DecisionTreeClassifier(max_depth=3, random_state=100)

dt_model.fit(x_sample, y_sample)
dt_pred = dt_model.predict(xuetangx_X_test)

print("Decision Tree Accuracy:", accuracy_score(xuetangx_y_test, dt_pred))
print("\n", classification_report(xuetangx_y_test, dt_pred))

dt_cv = cross_val_score(dt_model, x_sample, y_sample, cv=5, n_jobs=1)
print("Decision Tree CV Mean:", np.mean(dt_cv))
print("Decision Tree CV Std:", np.std(dt_cv))
```

```python
knn_model = KNeighborsClassifier(n_neighbors=3)

knn_model.fit(x_sample, y_sample)
knn_pred = knn_model.predict(xuetangx_X_test)

print("kNN Accuracy:", accuracy_score(xuetangx_y_test, knn_pred))
print("\n", classification_report(xuetangx_y_test, knn_pred))

knn_cv = cross_val_score(knn_model, x_sample, y_sample, cv=5, n_jobs=1)
print("kNN CV Mean:", np.mean(knn_cv))
print("kNN CV Std:", np.std(knn_cv))
```

```python
lr_model = LogisticRegression(max_iter=1000, random_state=100)

lr_model.fit(x_sample, y_sample)
lr_pred = lr_model.predict(xuetangx_X_test)

print("LogReg Accuracy:", accuracy_score(xuetangx_y_test, lr_pred))
print("\n", classification_report(xuetangx_y_test, lr_pred))

lr_cv = cross_val_score(lr_model, x_sample, y_sample, cv=5, n_jobs=1)
print("LogReg CV Mean:", np.mean(lr_cv))
print("LogReg CV Std:", np.std(lr_cv))
```

```python
lda_model = LinearDiscriminantAnalysis()

lda_model.fit(x_sample, y_sample)
lda_pred = lda_model.predict(xuetangx_X_test)

print("LDA Accuracy:", accuracy_score(xuetangx_y_test, lda_pred))
print("\n", classification_report(xuetangx_y_test, lda_pred))

lda_cv = cross_val_score(lda_model, x_sample, y_sample, cv=5, n_jobs=1)
print("LDA CV Mean:", np.mean(lda_cv))
print("LDA CV Std:", np.std(lda_cv))
```

```python
nb_model = GaussianNB()

nb_model.fit(x_sample, y_sample)
nb_pred = nb_model.predict(xuetangx_X_test)

print("Naive Bayes Accuracy:", accuracy_score(xuetangx_y_test, nb_pred))
print("\n", classification_report(xuetangx_y_test, nb_pred))

nb_cv = cross_val_score(nb_model, x_sample, y_sample, cv=5, n_jobs=1)
print("NB CV Mean:", np.mean(nb_cv))
print("NB CV Std:", np.std(nb_cv))
```
